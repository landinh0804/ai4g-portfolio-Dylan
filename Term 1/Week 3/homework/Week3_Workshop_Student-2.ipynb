## Learning Wave 1 — Lists

### 🧠 First: what *is* a list?

Up to now, every value needed its own variable. That is fine for a handful of things — but imagine a food bank with 100 supplies. 100 separate variable names would be madness.

A **list** is **one variable that holds many things, in a row, in order.** Everyday lists you already know:

- 🛒 a **shopping list** — items written one under another, in the order you wrote them
- 🚂 a **train** — carriages 1, 2, 3… each holding something, in a fixed order
- 🥚 an **egg carton** — a row of slots, each holding one egg at a fixed position

You fetch an item by its **position**, called its **index**. The thing every beginner trips over: **Python starts counting at 0.**

```python
supplies = ["rice", "pasta", "beans", "lentils"]
#             0        1        2         3       <- index
#                                        -1       <- the last item

supplies[0]     # 'rice'    (first item)
supplies[-1]    # 'lentils' (last item, counted from the end)
len(supplies)   # 4         (how many items there are)
```

**Mental model:** Week 1 was *one box → one value*. A list is *one box → many values, in a numbered row.*

That is the whole idea. The exercise below just practises putting things in, taking things out, and looking things up.

### Exercise 2: Wage Gap Analyzer
⏱ **15 minutes** | Slide 10

**Context:** Across the EU, women earn on average about **12% less per hour** than men (Eurostat). You will measure the gap in a (simplified) salary sample of one company — the analysis pattern is exactly how real pay-gap audits start.

**Given:**
```python
salaries_men = [3800, 2950, 4200, 3100, 5050, 3400]
salaries_women = [3300, 2700, 3650, 2850, 4400, 2900]
```

**Steps:**
1. Calculate the average of each list: `sum(xs) / len(xs)`.
2. Calculate the gap: `(avg_men - avg_women) / avg_men * 100`, rounded to 1 decimal.
3. Combine both lists with `+` and print the overall `min()` and `max()`.
4. **Bonus:** print all salaries below the overall average — a `for` loop over the combined list with an `if` (Week 2 meets Week 3!).

````python

salaries_men = [3800, 2950, 4200, 3100, 5050, 3400]
salaries_women = [3300, 2700, 3650, 2850, 4400, 2900]

# Step 1: averages with sum() / len()
avg_men = sum(salaries_men) / len(salaries_men)
avg_women = sum(salaries_women) / len(salaries_women)

print("Average Men Salary:", avg_men)
print("Average Women Salary:", avg_women)

# Step 2: gap percentage, rounded to 1 decimal
gap = ((avg_men - avg_women) / avg_men) * 100
print(f"Wage Gap: {round(gap, 1)}%")

# Step 3: combine with + , print min() and max()
all_salaries = salaries_men + salaries_women
print("Minimum Salary:", min(all_salaries))
print("Maximum Salary:", max(all_salaries))

# Step 4 (Bonus): loop over the combined list, print salaries below the overall average
overall_avg = sum(all_salaries) / len(all_salaries)
print(f"Overall Average: {overall_avg:.2f}")
print("Salaries below the overall average:")

for salary in all_salaries:
    if salary < overall_avg:
        print(salary)
````

## Learning Wave 3 — Dictionaries

### 🧠 First: what *is* a dictionary?

A list is perfect when **order** matters and you fetch things by position. But you do not think of a friend as *contact number 7* — you think of them by **name**. That is exactly what a dictionary is for.

A **dictionary** stores **pairs**: a **key** (a label you already know) and the **value** it points to. Everyday dictionaries you already use:

- 📖 a **real dictionary** — look up a *word* (key) to get its *meaning* (value)
- 📇 **phone contacts** — look up a *name* (key) to get a *number* (value)
- 🧥 a **coat check** — hand in a *ticket* (key) to get your *coat* (value) back

```python
services = {"Centrum": 14, "Laak": 3, "Escamp": 6}
#             ^key   ^value

services["Laak"]              # 3   (look up by name, not by position)
services.get("Ypenburg", 0)   # 0   (safe: returns 0 instead of crashing)
```

**List vs. dictionary — the one line to remember:**

- a **list** answers *"give me the 3rd item"* (by **position**)
- a **dictionary** answers *"give me the item called Laak"* (by **name**)

Rule of thumb: order or sequence → use a **list**; look something up by a label → use a **dictionary**.

The exercise below builds a real dictionary — and shows what happens when something is *missing* from it.

### Exercise 3: City Services Map
⏱ **20 minutes** | Slide 15

**Context:** Where you live decides how close care is. Below: the number of health & social services per district of The Hague (simplified). A dictionary maps **keys** (districts) to **values** (service counts).

**Steps:**
1. Create the dictionary:
```python
services = {"Centrum": 14, "Escamp": 6, "Laak": 3, "Scheveningen": 9, "Segbroek": 8, "Loosduinen": 5}
```
2. Look up `'Laak'` with `[]`. Then look up `'Ypenburg'` with `.get('Ypenburg', 0)` — why does `services['Ypenburg']` crash but `.get()` not?
3. Loop over `services.items()` and print `'District X has Y services'` for each district.
4. Find the **most underserved district** (lowest value) — use the comparison-variable pattern from Week 2.
5. Ask the user for their district with `input()`; if it is `in` the dictionary, print its count, otherwise print that the district is **not in the data**.
6. **Reflection** (markdown cell below).

````python

# Step 1: create the services dictionary
services = {"Centrum": 14, "Laak": 3, "Escamp": 6}

# Step 2: look up 'Laak' with [] and 'Ypenburg' with .get()
laak_count = services["Laak"]
ypenburg_count = services.get("Ypenburg", 0)

print(f"Services in Laak: {laak_count}")
print(f"Services in Ypenburg: {ypenburg_count}")

# Step 3: loop over services.items()
print("\nAll district services:")
for district, count in services.items():
    print(f"- {district}: {count} services")

# Step 4: find the district with the LOWEST count
lowest_district = ''
lowest_count = 999

for district, count in services.items():
    if count < lowest_count:
        lowest_count = count
        lowest_district = district

print(f"\nDistrict with lowest services: {lowest_district} ({lowest_count})")

# Step 5: ask the user for their district, check with in
user_district = input("\nEnter your district name: ").strip()

if user_district in services:
    print(f"{user_district} has {services[user_district]} services logged.")
else:
    print(f"Sorry, {user_district} is not listed in the services map.")
````

### 🥫 Food Bank Manager

**Goal:** Build a menu-driven program that a food bank volunteer could actually use to manage stock — your Exercise 1 list, all grown up.

**Context:** Food banks run on volunteers and spreadsheets. A small tool that tracks stock and warns before items run out makes a real difference.

#### Start with this inventory:
```python
inventory = {"pasta": 20, "rice": 15, "beans": 30, "soup": 12, "cooking oil": 8}
```

#### Requirements:

| # | Requirement | Concepts |

| 1 | A menu in a `while True` loop: **1)** show stock **2)** add donation **3)** hand out package **4)** report **5)** quit — validate the choice (Week 2 pattern) | `while True`, validation |
| 2 | A function `show_stock(inventory)` that prints every item as a bar: `rice : ############### (15)` | `items()`, `'#' * qty` |
| 3 | Add donation: ask for item + quantity; if the item exists add to it, if not create it — use the counting pattern `inventory[item] = inventory.get(item, 0) + qty` | `get()`, counting pattern |

#### Bonus:
- Keep a list `families_served`: ask for a (first) name at every hand-out, `append()` it, and show `len(families_served)` in the report.
- A **shortage list**: in the report, build a list of all items below 5 and print it.
- **Reflection** (markdown cell below).

#### Tips:
1. Build one menu option at a time — test it — then the next.
2. `show_stock` first: you'll use it to debug everything else.
3. Item names from `input()` are case-sensitive: `'Rice'` ≠ `'rice'`. Can you fix that? (`.lower()` is a preview of next week!)
4. The exact layout of the bar chart is cosmetic — readable beats pretty.

````python
# Big Assignment: Food Bank Manager
inventory = {"pasta": 20, "rice": 15, "beans": 30, "soup": 12, "cooking oil": 8}

# --- Step 1: define show_stock(inventory) ---
def show_stock(inv):
    print("\n--- Current Inventory ---")
    for item, qty in inv.items():
        print(f"  {item.capitalize()}: {qty}")
    print("-------------------------")

# --- Step 2: the menu loop ---
while True:
    print("\n--- Food Bank Manager ---")
    print("1) Show stock  2) Add donation  3) Hand out package  4) Report  5) Quit")
    choice = input("Choice: ").strip()

    if choice == '1':
        show_stock(inventory)

    elif choice == '2':
        item = input("Enter item name to donate: ").strip().lower()
        try:
            qty = int(input("Enter quantity: "))
            if qty > 0:
                # Add to existing quantity or set new item
                inventory[item] = inventory.get(item, 0) + qty
                print(f"Successfully added {qty} x '{item}'.")
            else:
                print("Quantity must be greater than zero.")
        except ValueError:
            print("Invalid input! Quantity must be a whole number.")

    elif choice == '3':
        item = input("Enter item name to hand out: ").strip().lower()
        if item in inventory:
            try:
                qty = int(input(f"Enter quantity to hand out (Available: {inventory[item]}): "))
                if qty <= 0:
                    print("Quantity must be greater than zero.")
                elif qty <= inventory[item]:
                    inventory[item] -= qty
                    print(f"Successfully handed out {qty} x '{item}'.")
                    # Optionally remove item if stock reaches 0
                    if inventory[item] == 0:
                        del inventory[item]
                        print(f"'{item}' is now completely out of stock!")
                else:
                    print(f"Not enough stock! Only {inventory[item]} available.")
            except ValueError:
                print("Invalid input! Quantity must be a whole number.")
        else:
            print(f"'{item}' is not in the inventory.")

    elif choice == '4':
        print("\n--- Inventory Report ---")
        total_items = sum(inventory.values())
        print(f"Total items in stock: {total_items}")
        print(f"Unique product categories: {len(inventory)}")

        # Identify low stock items (less than 10)
        low_stock = [item for item, qty in inventory.items() if qty < 10]
        if low_stock:
            print(f"Low stock items (< 10): {', '.join(low_stock)}")
        else:
            print("All items are sufficiently stocked!")
        print("------------------------")

    elif choice == '5':
        print("Exiting Food Bank Manager. Goodbye!")
        break

    else:
        print("Invalid choice — enter 1-5.")
````

## 5 Things to Remember

1. **Lists keep order; indexes start at 0** — `supplies[0]` is the first item, `supplies[-1]` the last.
2. **`in` answers "is it there?"** — `'rice' in supplies`, `'Laak' in services` — one check, no loop needed.
3. **`sum()`, `len()`, `min()`, `max()` do the heavy lifting** — `average = sum(xs) / len(xs)`.
4. **Dictionaries map keys to values** — `inventory['rice']` crashes on a missing key; `inventory.get('rice', 0)` never does. The counting pattern `d[k] = d.get(k, 0) + n` is everywhere in data work.
5. **Data structures carry world views** — whoever is missing from your list or dict does not exist for your program. The same is true for AI training data.

### 🧠 First: what *is* an API?

All day you called functions: `len(supplies)`, `services.get('Laak', 0)`. An **API** is a function that runs on **someone else's computer**. You send it a request over the internet, it does the work, and it sends the answer back. The Gemini API and the Claude API are exactly that — a language model you call from Python:

```
your Python  ──▶  a question  ──▶  the model (on Google's or Anthropic's computers)
your Python  ◀──  an answer   ◀──
```

And look at the shape of it. The request you send is a **dictionary** (`{"role": "user", "content": "..."}`), a conversation is a **list of those dictionaries**, and the answer comes back as a **list** too. That is why this sits at the end of Week 3: you already speak its data format.
